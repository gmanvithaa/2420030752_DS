```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
p=0.3
data = np.random.binomial(n=1, p=p, size=1000)
sns.histplot(data, stat='density',discrete=True)
plt.title(f'Binomial Distribution (n=1, p={p})')
plt.xlabel('Number of Successes')
plt.ylabel('Density')
plt.xticks([0, 1])
plt.show()
```
